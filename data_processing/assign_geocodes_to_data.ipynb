### **Script to assign geocodes to data from 2021-2025!**

```python
import pandas as pd
import re

# ---------- Helper ----------
def normalize_address(addr):
    if pd.isna(addr):
        return None
    addr = str(addr).strip()
    addr = addr.strip('"').strip("'")
    addr = addr.replace(" ,", ",")
    addr = " ".join(addr.split())
    return addr


# ---------- Load cache once ----------
geo_df = pd.read_csv("../../data/cleaned_data/geocode_cache.csv")
geo_df["address_norm"] = geo_df["address"].apply(normalize_address)


# ---------- Loop over years ----------
for year in range(2021, 2026):

    print(f"\nProcessing year: {year}")

    # Load data
    file_path = f"../../data/cleaned_data/cleaned_ATL311_{year}.csv"
    requests_df = pd.read_csv(file_path)

    # Normalize
    requests_df["address_norm"] = requests_df["Address"].apply(normalize_address)

    # Merge
    merged_df = requests_df.merge(
        geo_df[["address_norm", "lat", "lng", "ok"]],
        on="address_norm",
        how="left"
    )

    # Split
    matched_df = merged_df[merged_df["lat"].notna()]
    unmatched_df = merged_df[merged_df["lat"].isna()]

    # Stats
    total = len(merged_df)
    unmatched = len(unmatched_df)
    matched = len(matched_df)

    unmatched_percentage = (unmatched / total) * 100
    matched_percentage = (matched / total) * 100

    print(f"Total: {total}")
    print(f"Matched: {matched} ({matched_percentage:.2f}%)")
    print(f"Unmatched: {unmatched} ({unmatched_percentage:.2f}%)")

    # Save output
    output_path = f"../../data/cleaned_data/atl311_{year}_merged_with_geocodes.csv"
    merged_df.to_csv(output_path, index=False)

    print(f"Saved: {output_path} ✅")
```

```python
files = {
    2021: "../../data/cleaned_data/atl311_2021_merged_with_geocodes.csv",
    2022: "../../data/cleaned_data/atl311_2022_merged_with_geocodes.csv",
    2023: "../../data/cleaned_data/atl311_2023_merged_with_geocodes.csv",
    2024: "../../data/cleaned_data/atl311_2024_merged_with_geocodes.csv",
    2025: "../../data/cleaned_data/atl311_2025_merged_with_geocodes.csv",
}

dfs = []

for year, path in files.items():
    df = pd.read_csv(
        path,
        parse_dates=["Opened", "Closed Date 1", "Closed Date 2", "closed"],
        low_memory=False
    )

    print(f"{year}: raw rows = {len(df):,}")

    # Keep only valid geocodes
    df = df[
        df["ok"]
        .astype(str)
        .str.strip()
        .str.lower()
        .isin(["true", "1", "yes"])
    ]

    df = df[
        df["lat"].between(33.60, 33.90) &
        df["lng"].between(-84.55, -84.25)
    ].copy()

    # Add year for temporal analysis
    df["year"] = year

    print(f"{year}: after filter = {len(df):,}")

    dfs.append(df)
```

```python
df_all = pd.concat(dfs, ignore_index=True)
print(f"\nTotal rows (2021–2025): {len(df_all):,}")
df_all["year"].value_counts().sort_index()
df_all[["lat", "lng"]].describe()

df_all.to_csv("../../data/cleaned_data/merged_2021_2025.csv", index=False)
```
